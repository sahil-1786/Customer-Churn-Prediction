# **Phase 3: Data Cleaning & Preprocessing**

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Project_dataset.csv")
```

```python
df_clean = df.copy()
```

```python
df_clean.count()
```

```python
df_clean.drop("customerID", axis=1, inplace=True)

print("customerID column removed successfully.")
```

```python
print(df_clean)
```

```python
df_clean.head()
```

```python
df_clean["Churn"].value_counts()
```

```python
duplicates = df_clean.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
df_clean.drop_duplicates(inplace=True)
df_clean.count()
```

```python
missing = df_clean.isnull().sum()

missing = missing[missing > 0]

print(missing)
```

```python
print(df_clean["TotalCharges"].dtype)
```

```python
df_clean[df_clean["TotalCharges"] == " "]
```

```python
(df_clean["TotalCharges"] == " ").sum()
```

```python
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)
```

```python
df_clean.head()
```

```python
df_clean["TotalCharges"].dtype

# Notice the data type change from object to float
```

```python
(df_clean["TotalCharges"] == " ").sum()
```

```python
df_clean["TotalCharges"].isnull().sum()

# before 11 the
```

```python
median_charge = df_clean["TotalCharges"].median()

df_clean["TotalCharges"].fillna(median_charge, inplace=True)
```

```python
df_clean["TotalCharges"].isnull().sum()
```

```python
df_clean["Churn"] = df_clean["Churn"].map({
    "Yes":1,
    "No":0
})
```

```python

import matplotlib.pyplot as plt

for col in numerical_cols:
    plt.figure(figsize=(6,3))
    plt.boxplot(df_clean[col], vert=False)
    plt.title(col)
    plt.show()
```

```python
import seaborn as sns

sns.boxplot(x="Churn", y="tenure", data=df_clean)
```

# **Business Interpretation**



***"The boxplot shows a strong relationship between tenure and churn. Customers who remain with the company generally have much longer tenure, with a median of approximately 38 months. In contrast, customers who churn have a median tenure of only about 10 months. This suggests that the highest risk of churn occurs during the early months of a customer's relationship with the company. Therefore, retention strategies should focus on improving customer experience during the first year."***
