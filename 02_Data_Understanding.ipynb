```python
import pandas as pd
import numpy as np
```

```python
# ------------------------------------------
# Load Dataset
# ------------------------------------------

dataset_path = "/content/Project_dataset.csv"      # Change if necessary

df = pd.read_csv(dataset_path)
df
```

```python
# ------------------------------------------
# Display First & Last Records
# ------------------------------------------

print("="*60)
print("First 5 Rows")
print("="*60)
print(df.head())
```

```python
print("="*60)
print("Last 5 Rows")
print("="*60)
print(df.tail())
```

```python

# ------------------------------------------
# Shape of Dataset
# ------------------------------------------

print("\n")
print("="*60)
print("Dataset Shape")
print("="*60)

rows, cols = df.shape

print(f"Number of Rows    : {rows}")
print(f"Number of Columns : {cols}")
```

```python
# ------------------------------------------
# Column Names
# ------------------------------------------

print("\n")
print("="*60)
print("Column Names")
print("="*60)

print(df.columns.tolist())
```

```python
# ------------------------------------------
# Dataset Information
# ------------------------------------------

print("\n")
print("="*60)
print("Dataset Information")
print("="*60)

df.info()
```

```python
# ------------------------------------------
# Data Types
# ------------------------------------------

print("\n")
print("="*60)
print("Data Types")
print("="*60)

print(df.dtypes)
```

```python
# ------------------------------------------
# Missing Values
# ------------------------------------------

print("\n")
print("="*60)
print("Missing Values")
print("="*60)

missing = df.isnull().sum()

print(missing)

print("\nTotal Missing Values :", missing.sum())
```

```python
# ------------------------------------------
# Duplicate Records
# ------------------------------------------

print("\n")
print("="*60)
print("Duplicate Records")
print("="*60)

duplicates = df.duplicated().sum()

print("Total Duplicate Rows :", duplicates)
```

```python
# ------------------------------------------
# Numerical Summary
# ------------------------------------------

print("\n")
print("="*60)
print("Numerical Statistics")
print("="*60)

print(df.describe())
```

```python
# ------------------------------------------
# Categorical Summary
# ------------------------------------------

print("\n")
print("="*60)
print("Categorical Statistics")
print("="*60)

print(df.describe(include='object'))
```

```python
# ------------------------------------------
# Unique Values in Each Column
# ------------------------------------------

print("\n")
print("="*60)
print("Unique Values")
print("="*60)

for column in df.columns:
    print(f"{column:<25} {df[column].nunique()}")
```

```python

# ------------------------------------------
# Class Distribution
# ------------------------------------------

print("\n")
print("="*60)
print("Target Variable Distribution")
print("="*60)

print(df["Churn"].value_counts())

print("\n")

print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
# ------------------------------------------
# Memory Usage
# ------------------------------------------

print("\n")
print("="*60)
print("Memory Usage")
print("="*60)

memory = df.memory_usage(deep=True).sum() / (1024**2)

print(f"Dataset uses {memory:.2f} MB")
```
